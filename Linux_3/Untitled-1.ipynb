# Conda 安装 Node.js 
## 1. 前提
已安装 **Miniconda** 并打开终端（Linux终端）

## 2. 核心命令（一键执行）
```bash
# 1. 创建并激活专用环境（推荐，避免污染全局）
conda create -n node_env nodejs -y
conda activate node_env

# 2. 直接验证安装+启动 REPL
node
```

## 3. 关键验证
- 安装成功：终端显示 `Node.js vxx.x.x` 版本号
- 进入 REPL：出现 `>` 提示符，可直接执行 JS 代码（如 `2+3`、`console.log("hello")`）
- 退出 REPL：按 `Ctrl + C` 两次 或 输入 `.exit` 回车


### 总结
1. 用 conda 直接安装 `nodejs` 包即可
2. 输入 `node` 命令直接启动 JS REPL
3. 退出用 `Ctrl+C` 或 `.exit`
