```python
from collections.abc import Sequence
import random

class QuickSort:
    """
    快速排序类，包含分区和排序功能。
    支持：
    - 随机轴快速排序（原 sort 方法）
    - 三路快速排序（sort_3way 方法）
    """

    def __init__(self, arr: Sequence[int | float]):
        """
        初始化快速排序器。

        :param arr: 待排序的数组，可以是整数或浮点数
        """
        self.arr: list[int | float] = list(arr)
        self.length: int = len(arr)

    def partition(self, left_index: int, right_index: int) -> int:
        """
        对数组的指定范围进行分区操作（随机选择轴）。

        随机选择一个元素作为轴，将其交换到最右侧，
        然后使用双指针分区，将数组分为两部分：
        - 左边的元素都小于轴
        - 右边的元素都大于轴

        :param left_index: 分区范围的左边界索引
        :param right_index: 分区范围的右边界索引
        :return: 轴的最终位置索引
        """
        # 随机选择轴并与最右元素交换
        pivot_rand_index = random.randint(left_index, right_index)
        self.arr[pivot_rand_index], self.arr[right_index] = self.arr[right_index], self.arr[pivot_rand_index]

        pivot: int | float = self.arr[right_index]
        right_pointer: int = right_index - 1
        left_pointer: int = left_index

        while True:
            while left_pointer <= right_pointer and self.arr[left_pointer] < pivot:
                left_pointer += 1
            while left_pointer <= right_pointer and self.arr[right_pointer] > pivot:
                right_pointer -= 1
            if left_pointer >= right_pointer:
                break
            self.arr[left_pointer], self.arr[right_pointer] = self.arr[right_pointer], self.arr[left_pointer]

        # 将轴放到正确位置
        self.arr[left_pointer], self.arr[right_index] = self.arr[right_index], self.arr[left_pointer]
        return left_pointer

    def sort(
        self, left_index: int | None = None, right_index: int | None = None
    ) -> list[int | float]:
        """
        随机轴快速排序（原地递归）。

        :param left_index: 排序范围的左边界索引，默认为数组开头
        :param right_index: 排序范围的右边界索引，默认为数组结尾
        :return: 排序后的数组
        """
        if left_index is None:
            left_index = 0
        if right_index is None:
            right_index = self.length - 1

        if right_index - left_index <= 0:
            return self.arr

        pivot_position = self.partition(left_index, right_index)
        self.sort(left_index, pivot_position - 1)
        self.sort(pivot_position + 1, right_index)
        return self.arr

    # ---------- 三路快速排序 ----------
    def partition_3way(self, left_index: int, right_index: int):
        pivot_rand_index = random.randint(left_index, right_index)
        self.arr[pivot_rand_index], self.arr[left_index] = self.arr[left_index], self.arr[pivot_rand_index]
        pivot = self.arr[left_index]

        lt = left_index      
        i = left_index + 1
        gt = right_index     

        while i <= gt:
            if self.arr[i] < pivot:
                self.arr[lt], self.arr[i] = self.arr[i], self.arr[lt]
                lt += 1
                i += 1
            elif self.arr[i] > pivot:
                self.arr[i], self.arr[gt] = self.arr[gt], self.arr[i]
                gt -= 1
            else:  
                i += 1
        return lt, gt

    def sort_3way(
        self, left_index: int | None = None, right_index: int | None = None
    ) -> list[int | float]:
        """
        :param left_index: 排序范围的左边界索引
        :param right_index: 排序范围的右边界索引
        :return: 排序后的数组
        """
        if left_index is None:
            left_index = 0
        if right_index is None:
            right_index = self.length - 1

        if right_index - left_index <= 0:
            return self.arr

        lt, gt = self.partition_3way(left_index, right_index)
        # lt 指向第一个等于 pivot 的元素，gt 指向最后一个等于 pivot 的元素
        self.sort_3way(left_index, lt - 1)
        self.sort_3way(gt + 1, right_index)
        return self.arr

```

```python
if __name__ == "__main__":
    arr1 = [3, 6, 8, 10, 1, 2, 1]
    qs = QuickSort(arr1)
    print("随机轴排序结果:", qs.sort())

    arr2 = [5, 2, 5, 1, 5, 3, 5, 4, 5]
    qs2 = QuickSort(arr2)
    print("三路排序结果:  ", qs2.sort_3way())

    arr3 = list(range(10, 0, -1))
    qs3 = QuickSort(arr3)
    print("逆序数组排序:", qs3.sort())
```
