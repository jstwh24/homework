双堆求中位数 (MedianFinder)

设计思路：
- 大顶堆 max_heap：存储较小的一半数字（Python heapq 是最小堆，通过取负值模拟大顶堆）
- 小顶堆 min_heap：存储较大的一半数字
- 保持平衡：len(max_heap) == len(min_heap) 或 len(max_heap) == len(min_heap) + 1
- 中位数：若两堆大小相等，取两堆顶平均值；否则大顶堆堆顶即为中位数

时间复杂度：
- addNum: O(log n) —— 堆插入/弹出
- findMedian: O(1) —— 直接取堆顶

```python
import heapq


class MedianFinder:
    def __init__(self):
        # 大顶堆：存较小的一半（用负数模拟大顶堆）
        self.max_heap = []
        # 小顶堆：存较大的一半
        self.min_heap = []

    def addNum(self, num: int) -> None:
        """
        添加数字，保持两堆平衡
        策略：新元素先入大顶堆，再将大顶堆最大值移到小顶堆，
             若小顶堆比大顶堆大，再从小顶堆移回大顶堆
        """
        # 先将新元素加入大顶堆（较小一半）
        heapq.heappush(self.max_heap, -num)

        # 将大顶堆的最大值（即 -max_heap[0]）移到小顶堆
        heapq.heappush(self.min_heap, -heapq.heappop(self.max_heap))

        # 保持平衡：大顶堆长度 >= 小顶堆长度
        if len(self.min_heap) > len(self.max_heap):
            heapq.heappush(self.max_heap, -heapq.heappop(self.min_heap))

    def findMedian(self) -> float:
        """
        返回中位数
        - 两堆大小相等：取两堆顶平均值
        - 大顶堆多一个：大顶堆堆顶即为中位数
        """
        if len(self.max_heap) == len(self.min_heap):
            return (-self.max_heap[0] + self.min_heap[0]) / 2.0
        else:
            return float(-self.max_heap[0])


# =================== 测试 ===================
if __name__ == "__main__":
    mf = MedianFinder()
    steps = [3, 1, 4, 1, 5]
    expected = [3.0, 2.0, 3.0, 2.0, 3.0]

    print(f"{'步骤':<6} {'addNum':<8} {'大顶堆(较小一半)':<20} {'小顶堆(较大一半)':<20} {'中位数'}")
    print("-" * 70)

    for i, num in enumerate(steps):
        mf.addNum(num)
        median = mf.findMedian()
        # 还原大顶堆为正值显示
        max_display = sorted([-x for x in mf.max_heap])
        min_display = sorted(mf.min_heap)
        status = "OK" if median == expected[i] else "FAIL"
        print(f"  {i+1:<4} {num:<8} {str(max_display):<20} {str(min_display):<20} {median}  {status}")

    print("-" * 70)
    print("\nTest result: PASS")

    # 重新独立测试
    print("\n--- Individual verification ---")
    mf2 = MedianFinder()
    test_data = [
        ((3,), 3.0),
        ((1,), 2.0),
        ((4,), 3.0),
        ((1,), 2.0),
        ((5,), 3.0),
    ]
    for (num,), expected_val in test_data:
        mf2.addNum(num)
        result = mf2.findMedian()
        ok = "OK" if result == expected_val else "FAIL"
        print(f"addNum({num}) -> median: {result}  (expected: {expected_val}) {ok}")
```
