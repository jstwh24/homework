```python
from collections.abc import Sequence
from typing import Any

class SetResearchInsertDelete:
    """
    实现集合查找插入删除功能
    """

    def __init__(self, iterable: Sequence[Any]):
        """

        :param iterable:
        """
        self.iterable = list(iterable)

    def research(self, element: Any) -> bool:
        """
        查找功能
        :param element:
        :return:
        """

        for item in self.iterable:
            if item == element:
                return True
            return False

    def insert(self,index:int,element: Any) -> None:
        """
        插入功能
        :param index:
        :param element:
        :return:
        """

        if index >= len(self.iterable):
            print("index out of range")
        else:
            if self.research(element):
                print(f"{element} already in the set")
            else:
                self.iterable.insert(index,element)


    def delete(self,element: Any) -> None:
        """
        删除
        :param element:
        :return:
        """

        if self.research(element):
            self.iterable.remove(element)
        else:
            print(f"{element} not in the set")

```

```python
fruits = ['apple','banana','mango','cherry','orange']
set_insert = SetResearchInsertDelete(fruits)
set_insert.insert(2,'pear')
print(set_insert.iterable)
```

```python
from collections.abc import Sequence
from typing import Any

class OrderedArrayInsert:
    """
    有序数组的插入
    """
    def __init__(self, iterable: Sequence[Any]):
        """
        :param iterable: 有序数组初始化
        """
        self.iterable = list(iterable)
        self.len = len(self.iterable)

    def insert(self, element: Any) -> None:
        """
        插入元素并保持数组升序
        :param element: 待插入的元素
        """

        insert_index = self.len
        for index, value in enumerate(self.iterable):
            if element < value:
                insert_index = index
                break

        self.iterable.append(None)
        self.len += 1

        for i in range(self.len - 2, insert_index - 1, -1):
            self.iterable[i + 1] = self.iterable[i]

        self.iterable[insert_index] = element
```

```python
arr = OrderedArrayInsert([1, 3, 5, 7])
arr.insert(9)
print(arr.iterable)  # 输出: [1, 3, 4, 5, 7]
arr.insert(4)
print(arr.iterable)
```
