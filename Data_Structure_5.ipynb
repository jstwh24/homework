```python
class Node:
    def __init__(self, data: str | int | float):
        self.data: str | int | float = data
        self.next_node: Node | None = None

    def __str__(self) -> str:
        return f"Node({self.data!r})"


class LinkedList:
    def __init__(self):
        self.head: Node | None = None

    def append(self, data):
        """在链表末尾添加元素"""
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        last = self.head
        while last.next_node:
            last = last.next_node
        last.next_node = new_node

    def prepend(self, data):
        """在链表开头添加元素"""
        new_node = Node(data)
        new_node.next_node = self.head
        self.head = new_node

    def insert(self, index: int, data):
        """在指定索引位置插入元素"""
        if index == 0:
            self.prepend(data)
            return
        
        new_node = Node(data)
        current_node: Node | None = self.head
        current_index = 0
        
        while current_index < index - 1:
            if not current_node:
                raise IndexError("Index out of range")
            current_node = current_node.next_node
            current_index += 1
        
        if not current_node:
            raise IndexError("Index out of range")
        
        new_node.next_node = current_node.next_node
        current_node.next_node = new_node

    def __str__(self):
        nodes = []
        current = self.head
        while current:
            nodes.append(str(current.data))
            current = current.next_node
        return ' -> '.join(nodes)


def read(self, index: int) -> str | int | float | None:
    """读取指定索引位置的数据"""
    current_node: Node | None = self.head
    current_index = 0
    while current_index < index:
        if not current_node:
            return None  # 索引越界
        current_node = current_node.next_node
        current_index += 1
    return current_node.data if current_node and current_index == index else None


def delete(self, index: int) -> bool:
    """删除指定索引位置的节点"""
    if not self.head:
        return False
    
    if index == 0:
        self.head = self.head.next_node
        return True
    
    current_node: Node | None = self.head
    current_index = 0
    while current_index < index - 1:
        if not current_node or not current_node.next_node:
            return False
        current_node = current_node.next_node
        current_index += 1
    
    if current_node.next_node:
        current_node.next_node = current_node.next_node.next_node
        return True
    return False


def delete_by_value(self, value: str | int | float) -> bool:
    """删除第一个匹配值的节点"""
    if not self.head:
        return False
    
    if self.head.data == value:
        self.head = self.head.next_node
        return True
    
    current_node: Node | None = self.head
    while current_node.next_node:
        if current_node.next_node.data == value:
            current_node.next_node = current_node.next_node.next_node
            return True
        current_node = current_node.next_node
    return False


def length(self) -> int:
    """获取链表长度"""
    count = 0
    current = self.head
    while current:
        count += 1
        current = current.next_node
    return count


def find(self, value: str | int | float) -> int:
    """查找值的位置，返回索引，如果没找到返回-1"""
    current = self.head
    index = 0
    while current:
        if current.data == value:
            return index
        current = current.next_node
        index += 1
    return -1


def get_all_data(self) -> list[str | int | float]:
    """获取所有数据到列表中"""
    data_list = []
    current = self.head
    while current:
        data_list.append(current.data)
        current = current.next_node
    return data_list


# 将方法绑定到LinkedList类
LinkedList.read = read
LinkedList.delete = delete
LinkedList.delete_by_value = delete_by_value
LinkedList.length = length
LinkedList.find = find
LinkedList.get_all_data = get_all_data
```

```python
# 示例使用
if __name__ == "__main__":
    # 创建节点并连接
    node_1 = Node(1)
    node_2 = Node(2)
    node_3 = Node(3)
    node_1.next_node = node_2
    node_2.next_node = node_3

    # 创建链表
    list_1 = LinkedList()
    list_1.head = node_1

    print("=== 原始链表 ===")
    print(node_1)
    print(node_2)
    print(node_3)
    print("链表内容:", list_1)

    # 测试读取方法
    print("\n=== 测试读取功能 ===")
    print("读取索引0:", list_1.read(0))
    print("读取索引1:", list_1.read(1))
    print("读取索引2:", list_1.read(2))
    print("读取索引3:", list_1.read(3))  # 超出范围

    # 测试删除方法
    print("\n=== 测试删除功能 ===")
    print("删除前:", list_1)
    print("删除索引1的结果:", list_1.delete(1))
    print("删除后:", list_1)

    # 创建新链表测试其他功能
    list_2 = LinkedList()
    list_2.append("hello")
    list_2.append("world")
    list_2.append(42)
    list_2.append(3.14)

    print("\n=== 新链表测试 ===")
    print("新链表:", list_2)
    print("链表长度:", list_2.length())
    print("查找'world'的索引:", list_2.find("world"))
    print("查找999的索引:", list_2.find(999))

    # 测试插入功能
    list_2.insert(1, "inserted")
    print("在索引1插入'inserted':", list_2)

    # 测试按值删除
    print("删除值42的结果:", list_2.delete_by_value(42))
    print("删除后:", list_2)

    # 打印所有数据
    print("所有数据:", list_2.get_all_data())

    # 遍历链表
    print("\n=== 遍历链表 ===")
    current = list_2.head
    while current:
        print("节点数据:", current.data)
        current = current.next_node
```
