```python
# 定义单链表节点类
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val  # 节点值
        self.next = next  # 指向下一个节点的指针

# 链表操作类（包含反转、环检测、打印链表方法）
class LinkedList:
    def __init__(self):
        self.head = None  # 链表头节点

    # 1. 链表反转方法（核心：迭代法，空间复杂度O(1)）
    def reverse(self):
        prev = None  # 前驱节点，初始为None
        current = self.head  # 当前节点，从头节点开始
        
        while current:
            next_node = current.next  # 保存下一个节点（防止断链）
            current.next = prev  # 反转当前节点的指针，指向前驱
            prev = current  # 前驱节点后移
            current = next_node  # 当前节点后移
        
        # 循环结束后，prev就是新的头节点
        self.head = prev

    # 2. 检测链表是否有环（核心：快慢指针法/龟兔赛跑算法）
    def has_cycle(self):
        # 空链表或只有一个节点，无环
        if not self.head or not self.head.next:
            return False
        
        slow = self.head  # 慢指针，每次走1步
        fast = self.head  # 快指针，每次走2步
        
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            
            # 快慢指针相遇，说明有环
            if slow == fast:
                return True
        
        # 快指针走到链表尾部，无环
        return False

    # 辅助方法：向链表尾部添加节点
    def append(self, val):
        new_node = ListNode(val)
        if not self.head:
            self.head = new_node
            return
        last = self.head
        while last.next:
            last = last.next
        last.next = new_node

    # 辅助方法：打印链表（无环时使用）
    def print_list(self):
        res = []
        current = self.head
        while current:
            res.append(str(current.val))
            current = current.next
        print(" -> ".join(res))

# ------------------- 测试用例 -------------------
if __name__ == "__main__":
    # 测试1：链表反转
    print("=== 测试链表反转 ===")
    ll = LinkedList()
    ll.append('a')
    ll.append('b')
    ll.append('c')
    print("原链表：", end="")
    ll.print_list()  # 输出：a -> b -> c
    
    ll.reverse()
    print("反转后：", end="")
    ll.print_list()  # 输出：c -> b -> a

    print("\n=== 测试环检测 ===")
    # 测试2：无环链表
    ll2 = LinkedList()
    ll2.append(1)
    ll2.append(2)
    ll2.append(3)
    print("无环链表检测结果：", ll2.has_cycle())  # 输出：False

    # 测试3：有环链表（手动构造环）
    ll3 = LinkedList()
    ll3.append(1)
    ll3.append(2)
    ll3.append(3)
    # 让最后一个节点的next指向第一个节点，构造环
    ll3.head.next.next.next = ll3.head
    print("有环链表检测结果：", ll3.has_cycle())  # 输出：True
```
